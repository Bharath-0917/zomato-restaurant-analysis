```python
df_raw = df.copy()
```

```python
df.columns = df.columns.str.strip().str.replace(" ", "_").str.lower()
df.columns
```

```python
# fix missing values


df["cuisines"] = df["cuisines"].fillna("Unknown")
df.isna().sum().sum()
```

```python
# remove duplicates

df = df.drop_duplicates()
df.shape
```

```python
country_map = {1: "India", 14: "Australia", 30: "Brazil", 37: "Canada",
               94: "Indonesia", 148: "New Zealand", 162: "Philippines",
               166: "Qatar", 184: "Singapore", 189: "South Africa",
               191: "Sri Lanka", 208: "Turkey", 214: "UAE",
               215: "United Kingdom", 216: "United States"}

df.columns = df.columns.str.strip().str.replace(" ", "_").str.lower()   # safe to run twice

df["country"] = df["country_code"].map(country_map)
df["country"].value_counts()
```

```python
for col in ["has_table_booking", "has_online_delivery", "is_delivering_now"]:
    df[col] = df[col].eq("Yes")

df[["has_table_booking", "has_online_delivery"]].head()
```

```python
#handle the zero rating



df["is_rated"] = df["aggregate_rating"] > 0
df["rating"] = df["aggregate_rating"].where(df["is_rated"])

print("Average with zeros   :", round(df["aggregate_rating"].mean(), 2))
print("Average without zeros:", round(df["rating"].mean(), 2))
```

```python
#handle the zero costs


import numpy as np

df["avg_cost_for_two"] = df["average_cost_for_two"].replace(0, np.nan)
```

```python
df = df.drop(columns=["switch_to_order_menu", "average_cost_for_two"], errors="ignore")
df.shape
```

```python
#check the result

print("Before:", df_raw.shape)
print("After :", df.shape)
df.head()
```

```python
import matplotlib.pyplot as plt

india = df[df["country"] == "India"]
rated = df[df["is_rated"]]
```

```python
country_counts = df["country"].value_counts()

plt.figure(figsize=(9, 5))
plt.barh(country_counts.index, country_counts.values, color="#cb202d")
plt.gca().invert_yaxis()              # biggest bar on top
plt.title("Restaurants per country")
plt.xlabel("Number of restaurants")
plt.show()
```

**Insight:** India has 8,652 of the 9,551 restaurants (about 90.6%), followed by the United States (434) and the United Kingdom (80). Most other countries have fewer than 60 restaurants, so the rest of the analysis focuses on India.

```python
country_counts = df["country"].value_counts()

plt.figure(figsize=(9, 5))
bars = plt.barh(country_counts.index, country_counts.values, color="#cb202d")
plt.gca().invert_yaxis()
plt.bar_label(bars, padding=3)          # writes the number at the end of each bar
plt.title("Restaurants per country")
plt.xlabel("Number of restaurants")
plt.savefig("restaurants_by_country.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(rated["rating"], bins=25, color="#cb202d", edgecolor="white")
plt.title("Distribution of ratings")
plt.xlabel("Rating")
plt.ylabel("Number of restaurants")
plt.show()
```

**Insight:** Ratings are roughly bell-shaped and centred around 3.4 (average 3.44, median 3.40). Half of the rated restaurants fall between 3.0 and 3.8, and very few score below 2.5 or above 4.5, so a rating above 4.0 is relatively uncommon. This chart includes only the 7,403 rated restaurants; the 2,148 "Not rated" restaurants are excluded.

```python
#top 10 cuisines in India


cuisine_rows = india.assign(cuisine=india["cuisines"].str.split(", ")).explode("cuisine")
top10 = cuisine_rows["cuisine"].value_counts().head(10)

plt.figure(figsize=(9, 5))
plt.barh(top10.index, top10.values, color="#cb202d")
plt.gca().invert_yaxis()
plt.title("Top 10 cuisines in India")
plt.xlabel("Number of restaurants")
plt.show()
```

**Insight:** North Indian (3,946), Chinese (2,690) and Fast Food (1,963) are by far the most common cuisines among Indian restaurants, followed by Mughlai (992). North Indian alone appears in more than four times as many restaurants as the 5th-ranked cuisine, Bakery. The remaining cuisines (Continental, Italian, South Indian, Cafe, Desserts) each appear in roughly 600 to 730 restaurants. Because one restaurant can list several cuisines, these counts add up to more than the number of restaurants.

```python
plt.figure(figsize=(9, 5))
bars = plt.barh(top10.index, top10.values, color="#cb202d")
plt.gca().invert_yaxis()
plt.bar_label(bars, padding=3)
plt.title("Top 10 cuisines in India")
plt.xlabel("Number of restaurants")
plt.savefig("top_cuisines_india.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
# average rating by price range



price = rated.groupby("price_range")["rating"].mean()

plt.figure(figsize=(7, 5))
plt.bar(price.index, price.values, color="#cb202d")
plt.ylim(2.5, 4.2)
plt.xticks(price.index)
plt.title("Average rating by price range")
plt.xlabel("Price range (1 = cheapest, 4 = priciest)")
plt.ylabel("Average rating")
plt.show()
```

**Insight:** Average rating rises with price range, from 3.24 for the cheapest restaurants (range 1) to 3.89 for the priciest (range 4). The biggest jump is between range 2 (3.38) and range 3 (3.78). This shows an association, not a cause: pricier restaurants may be rated higher because they are full-service places with better ambience, not because of the price itself. This chart covers rated restaurants across all countries.

```python
# votes vs rating (scatter plot)


plt.figure(figsize=(8, 5))
plt.scatter(rated["votes"], rated["rating"], alpha=0.3, color="#cb202d")
plt.xscale("log")
plt.title("Votes vs rating (log scale)")
plt.xlabel("Votes")
plt.ylabel("Rating")
plt.show()
```

**Insight:** Restaurants with more votes tend to have higher ratings. The correlation between votes (log scale) and rating is about 0.65, a fairly strong positive relationship. Restaurants with very few votes cluster at lower ratings (2.5 to 3.3), while restaurants with 1,000+ votes rarely score below 3.0. This does not show that votes cause good ratings: popular restaurants may attract more reviews because they are good, and well-rated restaurants are probably easier for new diners to discover and review. The log scale is used because a few restaurants have thousands of votes while most have far fewer.

```python
# restaurants per country

plt.figure(figsize=(9, 5))
sns.barplot(x=country_counts.values, y=country_counts.index, color="#cb202d")
plt.title("Restaurants per country")
plt.xlabel("Number of restaurants")
plt.show()
```

**Insight:** The dataset covers 15 countries, but it is heavily concentrated in India, which has 8,652 of the 9,551 restaurants (about 90.6%). The United States is second with 434, and the United Kingdom third with 80. Most other countries have only 20 to 60 restaurants, and Canada has just 4. Because of this imbalance, country comparisons would be unreliable, so the rest of the analysis focuses on India.

```python
# rating distribution

plt.figure(figsize=(8, 5))
sns.histplot(rated["rating"], bins=25, kde=True, color="#cb202d")
plt.title("Distribution of ratings")
plt.show()
```

**Insight:** Ratings are roughly bell-shaped, with most restaurants between 3.0 and 3.8 and a peak around 3.3 to 3.4 (average 3.44). The smooth curve shows a longer tail on the high side: few restaurants score above 4.5, so a rating of 4.5+ is rare. The chart covers the 7,403 rated restaurants and excludes the 2,148 "Not rated" ones.

```python
# average rating by price range


plt.figure(figsize=(7, 5))
sns.barplot(x=price.index, y=price.values, color="#cb202d")
plt.ylim(2.5, 4.2)
plt.title("Average rating by price range")
plt.xlabel("Price range")
plt.ylabel("Average rating")
plt.show()
```

**Insight:** Restaurants in higher price ranges get higher average ratings. The average rises from 3.24 in price range 1 to 3.38 in range 2, then jumps to 3.78 in range 3 and 3.89 in range 4. The biggest step (about 0.4 points) is between range 2 and range 3, so moving from mid-priced to premium restaurants is where ratings improve most. This is an association, not proof that price causes better ratings: premium restaurants usually offer more service, ambience and consistency. Note that the y-axis starts at 2.5, which makes the gaps look larger than they are.

```python
# votes vs rating

plt.figure(figsize=(8, 5))
sns.scatterplot(data=rated, x="votes", y="rating", alpha=0.3, color="#cb202d")
plt.xscale("log")
plt.title("Votes vs rating (log scale)")
plt.show()
```

**Insight:** Restaurants with more votes tend to have higher ratings. The correlation between votes (log scale) and rating is about 0.65, a fairly strong positive relationship. Restaurants with fewer than about 30 votes cluster at ratings of 2.5 to 3.3, while those with 1,000+ votes rarely score below 3.0. This is an association, not proof of cause: popular restaurants may earn more votes because they are good, and highly rated places are easier for new diners to find and review. A log scale is used on the x-axis because a few restaurants have thousands of votes while most have far fewer.

## Conclusion

This analysis of 9,551 Zomato restaurants across 15 countries found:

1. **The data is mostly India.** About 90.6% of restaurants (8,652) are in India, so the analysis focuses on the Indian market. Other countries have too few restaurants for reliable comparisons.

2. **A few cuisines dominate.** North Indian (3,946), Chinese (2,690) and Fast Food (1,963) are the most common cuisines in India.

3. **Higher price means higher ratings.** Average rating rises from 3.24 in price range 1 to 3.89 in price range 4, with the biggest jump between range 2 and range 3.

4. **Popular restaurants are rated higher.** The correlation between votes (log scale) and rating is about 0.65, so restaurants with more votes tend to score better.

**Data note:** 2,148 restaurants (22.5%) had a rating of 0, which means "Not rated". They were excluded from rating averages. Including them would have lowered the average from 3.44 to 2.67.

**Limitations:** These results show associations, not causes. The data is a single snapshot with no dates, and costs are in different currencies, so cost was not compared across countries.
